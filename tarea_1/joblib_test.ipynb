```python
from joblib import Parallel, delayed
import math
import time

def calcular_raiz(numero):
    """Calcula la raíz cuadrada y simula un trabajo más pesado."""
    return math.sqrt(numero)

numeros = range(1_000_000)

tareas = (delayed(calcular_raiz)(i) for i in numeros)
```

```python
with Parallel(n_job=2, verbose=5, batch_size=100, backend="loky") as parallel_pool:
    parallel_result = parallel_pool(tareas)
```

```python
with Parallel(n_job=2, verbose=1000, batch_size=5, backend="threading") as parallel_pool:
    parallel_result = parallel_pool(tareas)
```
